# Programación orientada a objetos con fracciones

Los objetos son construcciones especiales de programación que permiten hacer operaciones complejas.

Podemos cargar objetos de fracciones usando:
```
from fractions import Fraction
```
Este código trae los objetos <tt>Fraction</tt> de la biblioteca <tt>fractions</tt> de Python. Una vez que se ejecuta este código, no hace falta repetirlo.

Ahora podemos crear fracciones sencillas:
```
a = Fraction(1, 2)
```
Con este código, a=1/2.

Podemos hacer operaciones matemáticas sencillas con fracciones:
```
b = a + a
```
Esta línea suma 1/2 + 1/2.

Podemos imprimir el resultado:
```
print(b)
```

````python
# Importa Fraction de la biblioteca fractions:
from fractions import Fraction

# Crea la fracción 1/2
a = Fraction(1, 2)

# Podemos sumar las fracciones y ver el resultado:
b = a+a
print(b)
````

# Operaciones complejas con fracciones
Podemos hacer operaciones complejas como se muestra a continuación:
```
# Esto tiene formato de código
a = Fraction(1,2)  # a = 1/2
b = Fraction(1,3)  # b = 1/3
c = Fraction(1,9)
d = 3*(a-b)/c      # expresión compleja
print(d)
```




````python
a = Fraction(1,2)  # a = 1/2
b = Fraction(1,3)  # b = 1/3
c = Fraction(1,9)
d = 3*(a-b)/c      # expresión compleja
print(d)
````
